```python
my_table=doc.find("table", class_="wikitable sortable plainrowheaders")


th_tags = my_table.find_all('th')
names = []
for elem in th_tags:
  #finding the < a > tag
    a_links = elem.find_all("a")
    #getting the text inside the < a > tag
    for i in a_links:
        names.append(i.string)
print(names)
```

```python
final_list = names[9: ]
states = []
for str in final_list:
    if len(str) > 3:
        states.append(str)
print(states)
```

```python
def sayHello():
  print("Hello from us")
```

```python
import threading
class OurThread(threading.Thread):
  def __init__(self,name):
    super().__init__()
    self.name=name
  def run(self):
    sayHello()
t1 = OurThread("t1")
t2 = OurThread("t2")
t1.start()
t2.start()
```

```python
t1= threading.Thread(target=sayHello)
t2= threading.Thread(target=sayHello)
t1.start()
t2.start()
```

```python
import concurrent.futures
pool=concurrent.futures.ThreadPoolExecutor(max_workers=2)
pool.submit(sayHello)
pool.submit(sayHello)
pool.submit(sayHello)
pool.shutdown(wait=True)
```

```python
import threading
import time
lock=threading.Lock()
counter=0
def increment():
    print("Inside thread",threading.current_thread().getName())
    lock.acquire()
    try:
      global counter
      print("Inside increment",threading.current_thread().getName())
      time.sleep(3)
      counter+=1
      print(counter)
    except Exception as e:
      print("Exception",e)
    finally:
      lock.release()
import concurrent.futures
pool=concurrent.futures.ThreadPoolExecutor(max_workers=2)
pool.submit(increment)
pool.submit(increment)
pool.submit(increment)
pool.submit(increment)
pool.shutdown(wait=True)
```
