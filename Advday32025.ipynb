```python
from flask import Flask,request

app =Flask(__name__)

@app.route("/calculate")
def calculate():
    n1=request.args.get("n1",type=int)
    n2=request.args.get("n2",type=int)
    operator=request.args.get("op")
    if operator == "+":
        return str(n1+n2)
    elif operator == "-":
        return str(n1-n2)
    elif operator == "*":
        return str(n1*n2)
    elif operator =="/":
        return str(n1/n2)
    else:
        return "Something wrong"

app.run()
```

```python
from flask import Flask,request,jsonify

app =Flask(__name__)

empList={}

@app.route("/emp",methods=["GET"])
def getemps():
    params=request.args
    if "empid" in params:
        return jsonify(empList.get(params.get("empid")))
    else:
        return jsonify(empList)
@app.route("/add",methods=["POST"])
def addEmp():
    data=request.json
    empid=data.get("empid")
    empList[empid]=data
    return jsonify({"success":True})
    
@app.route("/update/<empid>",methods=["PUT"])
def updateEmp(empid):
    empid =int(empid)
    if empid in empList:
        empList[empid]=request.json
        return jsonify({"success":True})
    else:
        return jsonify({"success":False})
 
@app.route("/delete/<empid>",methods=["DELETE"])
def deleteEmp(empid):
    empid =int(empid)
    if empid in empList:
        del empList[empid]
        return  jsonify({"success":True})
    else:
         return jsonify({"success":False})
app.run()
```

```python
from sqlalchemy import ForeignKey
from sqlalchemy.orm import relationship
from sqlalchemy import create_engine,Column,Integer,String,MetaData,Table
from sqlalchemy.orm import sessionmaker
from sqlalchemy.ext.declarative import declarative_base
url="sqlite:///test.db"
enginee=create_engine(url)
Session=sessionmaker(bind=enginee)
session=Session()
Base=declarative_base()
class User(Base):
    __tablename__="user_account"
    id=Column(Integer,primary_key=True)
    name=Column(String(50))
    address=relationship("Address",back_populates="user")

    def __repr__(self):
        return str(id)

class Address(Base):
    __tablename__="address"
    id=Column(Integer,primary_key=True)
    email=Column(String(50))
    user_id=Column(Integer,ForeignKey("user_account.id"))
    user=relationship("User",back_populates="address")
    
    def __repr__(self):
        return str(id)
Base.metadata.create_all(enginee)
```

```python
user=User(id=1,name="Raj")
add1=Address(id=1,email="raj@gmail.com",user=user)
add2=Address(id=2,email="raj1@gmail.com",user=user)
session.add_all([user,add1,add2])
session.commit()
```
