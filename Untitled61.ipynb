```python
class Calculator:
    def __init__(self,a,b):
        self.a=a
        self.b=b
        self.result=0
    def add(self):
        self.result=self.a+self.b
    def sub(self):
        self.result=self.a-self.b
    def mul(self):
        self.result=self.a*self.b
    def divide(self):
        self.result=self.a/self.b
```

```python
#Inheritance
#2 classes share parent child relationship
```

```python
class Parent1:
    def __init__(self,n1):
        self.n1=n1
    def displayN1(self):
        print("n1",self.n1)
```

```python
class Child1(Parent1):
    def __init__(self,n1,n2):
        Parent1.__init__(self,n1)
        self.n2=n2
    
```

```python
c1=Child1(22,33)
```

```python
c1.n1
```

```python
c1.n2
```

```python
c1.displayN1()
```

```python
class Parent3:
    def __init__(self,n1):
        self.n1=n1
    def display(self):
        print("n1 from Parent3",self.n1)
class Parent4:
    def __init__(self,n2):
        self.n2=n2
    def display(self):
        print("n2 from Parent4",self.n2)
class Child2(Parent3,Parent4):#Ambuguity is resolved by sequence of inheritance
    def __init__(self,n1,n2):
        Parent3.__init__(self,n1)
        Parent4.__init__(self,n2)       
```

```python
c1=Child2(22,33)
```

```python
c1.display()
```

```python
from abc import ABC,abstractmethod
```

```python
class Vehicle(ABC):
    @abstractmethod
    def start_enginee(self):
        pass
```

```python
v1=Vehicle()
```

```python
class Car(Vehicle):
    def start_enginee(self):
        print("Enginee start from Car")
```

```python
c1=Car()
```

```python
c1.start_enginee()
```

```python
class PrivateEg:
    def __init__(self,n1,n2):
        self.n1=n1
        self.__n2=n2#private variable
    def displayn1(self):
        print("n1",self.n1)
    def display(self):
        print("n1-->",self.n1)
        print("n2-->",self.__n2)
    def __displayprivate(self):
        print("From private",self.n1,self.__n2)
```

```python
p1=PrivateEg(10,20)
```

```python
p1.n1
```

```python
p1.__n2
```

```python
p1.display()
```

```python
p1.displayn1()
```

```python
p1.__displayprivate()
```

```python
p1._PrivateEg__n2
```

```python
p1._PrivateEg__displayprivate()
```

```python
class CoffeMachine:
    def __init__(self):
        self._water_temp=40
    def _boil_water(self):
        print("Heating water",self._water_temp)
    def press(self):
        self._boil_water()
        print("Brewing coffee")
```

```python
class EspressoMachine(CoffeMachine):
    def boost_temperature(self):
        self._water_temp=100
        print("Brewing at",self._water_temp)
```

```python
e1=EspressoMachine()
```

```python
e1.boost_temperature()
```
