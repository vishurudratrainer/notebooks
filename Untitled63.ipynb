```python
#OOPL ---important concept is inheritance
```

```python
class Parent1:
    def __init__(self,n1):
        self.n1=n1
    def display(self):
        print("n1 from parent1",self.n1)
```

```python
class Child1(Parent1):#Child1 extends Parent1
    def __init__(self,n1,n2):
        #Base class constructor need to be called
        Parent1.__init__(self,n1)
        self.n2=n2
        
```

```python
c1 = Child1(22,33)
c1.display()
```

```python
print(c1.n1,c1.n2)
```

```python
class Parent2:
    def __init__(self,n1):
        self.n1=n1
    def display(self):
        print("n1 from Parent2",self.n1)
```

```python
class Parent3:
    def __init__(self,n2):
        self.n2=n2
    def display(self):
        print("n2 from Parent3",self.n2)
```

```python
class Child2(Parent2,Parent3):#Sequence of inheritance
    def __init__(self,n1,n2):
        Parent2.__init__(self,n1)
        Parent3.__init__(self,n2)
```

```python
c3=Child2(22,33)
```

```python
c3.display()
```

```python
class Child3(Parent3,Parent2):#Sequence of inheritance
    def __init__(self,n1,n2):
        Parent2.__init__(self,n1)
        Parent3.__init__(self,n2)
```

```python
c4=Child3(22,33)
```

```python
c4.display()
```

```python
#Operator overloading
#Complex ---real + i img
#add real1+real2 , img1+img2
```

```python
class Complex:
    def __init__(self,real,img):
        self.real=real
        self.img=img
    def __str__(self):
        return "Real {} Img {}".format(self.real,self.img)
    def add(self,c2):
        r1=self.real+c2.real
        i1=self.img+c2.img
        return Complex(r1,i1)
    def __add__(self,c2):
        r1=self.real+c2.real
        i1=self.img+c2.img
        return Complex(r1,i1)
```

```python
c1=Complex(20,30)
c2=Complex(10,40)
print(c1.add(c2))
```

```python
c3=c1+c2
```

```python
print(c3)
```

```python
class CalculatorUtil:
    @staticmethod
    def add(a,b):
        return a+b
    @staticmethod
    def sub(a,b):
        return a-b
    def mul(a,b):
        return a*b
```

```python
CalculatorUtil.add(22,33)
```

```python
class P1:
    def __init__(self,p1):
        self.p1=p1
    def displayP1(self):
        print("From P1",self.p1)
```

```python
class P2(P1):
    def __init__(self,p1,p2):
        P1.__init__(self,p1)
        self.p2=p2
    def displayP2(self):
        print("From P2",self.p1,self.p2)
```

```python
class P3:
    def __init__(self,p3):
        self.p3=p3
    def displayP3(self):
        print("From P3",self.p3)
```

```python
class C1(P2,P3):
    def __init__(self,p1,p2,p3):
        P2.__init__(self,p1,p2)
        P3.__init__(self,p3)
```

```python
c1=C1(20,30,40)
```

```python
c1.displayP1()
```

```python
c1.displayP2()
```

```python
c1.displayP3()
```
