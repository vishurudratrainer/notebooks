```python
def outer3(n):
    print("Within outer3 n",n)
    def inner(x):
        print("x is",x,"n is",n)
        return x**n
    return inner
```

```python
o1=outer3(4)
```

```python
o1(2)
```

```python
o1(3)
```

```python
#In python function can be passed as argument to another function
def outer4(param1):
    print("Within outer 4")
    def inner():
        print("Before calling param1")
        param1()
        print("After calling param1")
    return inner
    
```

```python
def hello():
    print("Hello from hello function")
```

```python
f1=outer4(hello)
```

```python
f1()
```

```python
@outer4#Thi is equaivalent of f1=outer4(hello)
def hello1():
    print("From hello1 ")
```

```python
hello1()
```

```python
@outer4
def welcome():
    print("Welcome")
```

```python
welcome()
```

```python
def factorial(n):
    if n == 1:
        return n
    else:
        return n*factorial(n-1)
```

```python
#Except
#try except
try:
    n1=int(input("Enter number"))
    print("n1",n1)
    n2=int(input("Enter number"))
    print("n2",n2)
    res =n1/n2
    print("res",res)
except ValueError as ve:
    print("Error in conversion",ve)
except Exception as e:
    print("Some error",e)
```

```python
try:
    password1=input("Enter password")
    password2=input("Enter password again")
    if password1 != password2:
        raise Exception("Passwords are not equal")
except Exception as e:
    print("Some exception",e)
```

```python
class PasswordNotEqual(Exception):
    pass
```

```python
try:
    password1=input("Enter password")
    password2=input("Enter password again")
    if password1 != password2:
        raise PasswordNotEqual("Passwords are not equal")
except PasswordNotEqual as pe:
    print("Password issue",pe)
except Exception as e:
    print("Some exception",e)
```

```python
#input as age
#age <0 or > 120  InvalidAgeException
```

```python
class AgeException(Exception):
    pass
    
```

```python
try:
    age=int(input("Enter age"))
    if age < 0 or age > 120:
        raise AgeException(str(age) +" is invalid")
except AgeException as ae:
    print("Age issue",ae)
except Exception as ex:
    print("Exception",ex)
```
