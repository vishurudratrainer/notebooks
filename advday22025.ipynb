```python
class Complex:
    def __init__(self,real,img):
        self.real=real
        self.img=img
    def add(self,other):
        real =self.real+other.real
        img=self.img+other.img
        return Complex(real,img)
    def __add__(self,other):
        real =self.real+other.real
        img=self.img+other.img
        return Complex(real,img)   
    def __sub__(self,other):
        real =self.real-other.real
        img=self.img-other.img
        return Complex(real,img) 
    def __str__(self):
        return "Real {} Img {}".format(self.real,self.img)
```

```python
class Point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __str__(self):
        return "x {} y{}".format(self.x,self.y)
    def __eq__(self,other):
        m1=self.x*self.x +self.y*self.y 
        m2=other.x*other.x+other.y*other.y
        print("equal")
        return m1==m2
    def __gt__(self,other):
        m1=self.x*self.x +self.y*self.y 
        m2=other.x*other.x+other.y*other.y
        print("greater")
        return m1>m2   
    def __ge__(self,other):
        m1=self.x*self.x +self.y*self.y 
        m2=other.x*other.x+other.y*other.y
        print("greate equal")
        return m1>=m2
    def __lt__(self,other):
        m1=self.x*self.x +self.y*self.y 
        m2=other.x*other.x+other.y*other.y
        print("less than")
        return m1<m2  
    def __le__(self,other):
        m1=self.x*self.x +self.y*self.y 
        m2=other.x*other.x+other.y*other.y
        print("less than equal")
        return m1<=m2   
```

```python
count=0
lock=threading.Lock()
def inc():
    global count
    print("Trying to get lock",threading.current_thread().name)
    time.sleep(2)
    lock.acquire()
    print("Lock is acquired",threading.current_thread().name)
    try:
        count=count+1
    except Exception as e:
        print("Exception",e)
    finally:
        lock.release()
        print("Lock is released",threading.current_thread())
        
```

```python
t1=threading.Thread(target=inc,name="thread1")
t2=threading.Thread(target=inc,name="thread2")
t3=threading.Thread(target=inc,name="thread3")
t1.start()
t2.start()
t3.start()
```

```python
import aiohttp
import asyncio
```

```python
async def download(session,url):
    async with session.get(url) as response:
        response.raise_for_status()
        return await response.text()
```

```python
async def downloadeg():
    urls=["https://jsonplaceholder.typicode.com/todos/1",
          "https://jsonplaceholder.typicode.com/todos/2",
          "https://jsonplaceholder.typicode.com/todos/3"]
    async with aiohttp.ClientSession() as session:
        tasks=[download(session,url) for url in urls]
        results=await asyncio.gather(*tasks)
        for url,content in zip(urls,results):
            print("url",url)
            print(content)
```

```python
def throwDice():
    return random.randint(1,6)
```

```python
score  = 0
while score != 30:
    diceValue=throwDice()
    if score+diceValue > 30:
        continue
    else:
        score=score+diceValue
        if score in location:
            if location[score]> score:
                print("Ladder")
            else:
                print("snake")
            score=location[score]
        print(score)
    
```

```python
def roll(playerkey,score):
    diceValue=throwDice()
    if score+diceValue > 30:
        print("non feasible")
    else:
        score=score+diceValue
        if score in location:
            if location[score]> score:
                print("Ladder")
                return location[score]
            else:
                print("snake")
                score=location[score]
    return(score)
    
```

```python
players={}
players["1"]=0
players["2"]=0
condition=1
while condition==1:
    for k,v in players.items():
        print(k,"rolling dice",v)
        score=roll(k,v)
        print(score)
        players[k]=score
        if players[k]==30:
            print(k,"won")
            condition=0
            break
```

```python
def playsnakeandladder(noofplayers):
    players={}
    for i in range(1,noofplayers+1):
        players[str(i)]=0

    condition=1
    while condition==1:
        for k,v in players.items():
            print(k,"rolling dice",v)
            score=roll(k,v)
            print(score)
            players[k]=score
            if players[k]==30:
                print(k,"won")
                condition=0
                break
```
