```python
class Parent1:
    def __init__(self,p1):
        self.p1=p1
    def displayP1(self):
        print("p1 from Parent1",self.p1)
        
```

```python
class Child1(Parent1):#Inheritance
    def __init__(self,p1,p2):
        Parent1.__init__(self,p1)#Calling the base class constructor
        self.p2=p2
    
```

```python
c1 = Child1(22,33)
```

```python
c1.p1
```

```python
c1.p2
```

```python
c1.displayP1()
```

```python
#Multiple Inheritance

#P1 display
#P2 display
#C -p1-p2--display
```

```python
class P1:
    def __init__(self,n1):
        self.n1=n1
    def display(self):
        print("From P1 n1=",self.n1)
```

```python
class P2:
    def __init__(self,n2):
        self.n2=n2
    def display(self):
        print("From P2 n2=",self.n2)
    
```

```python
class Child2(P1,P2):#In the case of conflict, resolves sequence of inheritance
    def __init__(self,n1,n2):
        P1.__init__(self,n1)
        P2.__init__(self,n2)
```

```python
c2 =Child2(22,33)
```

```python
c2.display()
```

```python
class Child3(P2,P1):
    def __init__(self,n1,n2):
        P1.__init__(self,n1)
        P2.__init__(self,n2)
```

```python
c3 =Child3(33,44)
```

```python
c3.display()
```

```python
class Shape(ABC):#Inheriting from ABC indicates abstract 
    @abstractmethod#We are saying that class which inherits has to give body..
    def area(self):
        pass #equivalent of no body
    def display(self):
        print("I am in Shape")
```

```python
#Operator overloading
class Complex:
    def __init__(self,real,img):
        self.real=real
        self.img = img
    def __str__(self):
        return "real {} img {}".format(self.real,self.img)
    def add(self,other):
        r1 =self.real+other.real
        i1 =self.img+other.img
        return Complex(r1,i1)
    def __add__(self,other):# implementing + operator
        r1 =self.real+other.real
        i1 =self.img+other.img
        return Complex(r1,i1)
```

```python
c1 =Complex(22,10)
c2 = Complex(20,40)
c3=c1.add(c2)
print(c3)
c4 =c1 +c2 #Example of operator overloading
print(c4)
```

```python
class Point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __lt__(self,other):
        print("less than is called")
        m1 =self.x*self.x+self.y*self.y
        m2 = other.x*other.x+other.y*other.y
        return m1 < m2
    def __le__(self,other):
        print("less than/equal is called")
        m1 =self.x*self.x+self.y*self.y
        m2 = other.x*other.x+other.y*other.y
        return m1 <= m2    
    def __gt__(self,other):
        print("greater than is called")
        m1 =self.x*self.x+self.y*self.y
        m2 = other.x*other.x+other.y*other.y
        return m1 > m2
    def __ge__(self,other):
        print("greater than/equal is called")
        m1 =self.x*self.x+self.y*self.y
        m2 = other.x*other.x+other.y*other.y
        return m1 >= m2
    def __eq__(self,other):
        print("equal  is called")
        m1 =self.x*self.x+self.y*self.y
        m2 = other.x*other.x+other.y*other.y
        return m1 == m2
              
```

```python
p1 =Point(1,2)
p2=Point(1,3)
```

```python
p1 < p2
```

```python
p1 > p2
```

```python
p1 <= p2
```

```python
p1 >=p2
```

```python
p1==p2
```

```python
#Meta programming
#Python
#Functions can be created within functions
#Functions can be passed as arguments to functions
#Functions can be returned as result
```

```python
def outer():
    print("Within outer")
    def inner():#Function can be created within function
        print("Within inner")
    inner()
```

```python
def outer1():
    print("Within outer1")
    def inner():#Function can be created within function
        print("Within inner")
    return inner #Functions can be returned from nother function
```

```python
f1 = outer1()#Storing result function in f1
```

```python
f1()
```

```python
outer1()()
```

```python
def outer2(param1):#Function can be passed as an argument
    print("Within outer before calling param")
    param1()
    print("Within outer after calling param")
```

```python
def hello():
    print("Hello from us")
```

```python
outer2(hello)
```

```python
def outer3(param1):
    print("Inside outer3")
    def inner():
        print("Inside inner")
        print("Before calling param1")
        param1()
        print("After calling param1")
        
    return inner
```

```python
f2 = outer3(hello)
```

```python
f2()
```

```python
#Generators
#are functions which are memory effective
#1000 numbers --generate one number use....
#We can create generator functions ---return ---yield
#yield is cpable of remembering value
```

```python
def generator1(n):
    i = 1
    while i <= n:
        yield i
        i = i+1
```
