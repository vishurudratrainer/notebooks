```python
import re

string = '39801 356 2102 1111'

# Three digit number followed by space followed by two digit number
pattern = '(\d{3}) (\d{2})'

# match variable contains a Match object.
match = re.search(pattern, string) 

if match:
  print(match.group())
else:
  print("pattern not found")

print(match.group(1))
print(match.group(2))

```

```python
#Ambuguity
#P1---display()---impl1
#P2 --display()--impl2
#Child p1,p2 display
```

```python
class P1:
    def __init__(self,n1):
        self.n1=n1
    def display(self):
        print("From Parent1",self.n1)
```

```python
class P2:
    def __init__(self,n2):
        self.n2=n2
    def display(self):
        print("From Parent2",self.n2)
```

```python
class Child(P1,P2):#Sequence of inheritance
    def __init__(self,n1,n2):
        P1.__init__(self,n1)
        P2.__init__(self,n2)
```

```python
c1=Child(22,33)
```

```python
c1.n1
```

```python
c1.n2
```

```python
c1.display()
```

```python
#Snake Ladder --OOOL
import random
class Board:
    def __init__(self,snakes,ladders,endingScore):
        self.snakes=snakes
        self.ladders=ladders
        self.score=0
        self.endingScore=endingScore

    def rollDice(self):
        return random.randint(1,6)
    def playGame(self):
        while self.score < self.endingScore:
            print("Rolling dice")
            dice=self.rollDice()
            print("Dice value",dice)
            nextScore=self.score + dice
            if nextScore in self.snakes:
                nextScore=snakes[nextScore]
                print("Snake bite")
            elif nextScore in self.ladders:
                nextScore=snakes[nextScore]
                print("LAdder jump")
            else:
                if nextScore <= self.endingScore:
                    self.score=nextScore
```

```python
board=Board({27:10,23:7},{11:24,16:29},30)
```

```python
board.playGame()
```

```python
class PasswordNotEqual(Exception):
    pass
```

```python
try:
    password1=input("Enter password1:")
    password2=input("Enter password2:")
    if password1 != password2:
        raise PasswordNotEqual("Password not equal")
except PasswordNotEqual as pe:
    print("PAssword not equal",pe)
except Exception as e:
    print("Error",e)
```
